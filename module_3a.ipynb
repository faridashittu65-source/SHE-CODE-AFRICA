```python
# Analysis with list

# Customer names
names = [
    "Amina Bello", "John Okeke", "Fatima Musa", "Chinedu Obi", "Sola Ade",
    "Maryam Yusuf", "Emeka Nwosu", "Oluchi Eze", "Uche Chukwu", "Kemi Adebayo",
    "Ibrahim Lawal", "Ngozi Ojo", "Tunde Bakare", "Rita Eze", "Daniel Okafor",
    "Halima Bello", "Segun Afolabi", "Blessing Chukwu", "Femi Ade", "Chika Nwankwo"
]

#  Account types
accounts = [
    "Savings", "Current", "Savings", "Fixed Deposit", "Savings",
    "Current", "Savings", "Fixed Deposit", "Current", "Savings",
    "Savings", "Current", "Fixed Deposit", "Savings", "Current",
    "Savings", "Current", "Savings", "Savings", "Fixed Deposit"
]

# Balances in Naira
balances = [
    250000, 120000, 54000, 1000000, 87500,
    300000, 450000, 1500000000, 200000, 95000,
    125000, 80000, 7500000, 60000, 180000,
    50000, 230000, 97000000, 11000000, 1300000
]

#  Branch cities
branches = [
    "Lagos", "Abuja", "Kano", "Enugu", "Ibadan",
    "Lagos", "Abuja", "Kano", "Enugu", "Ibadan",
    "Lagos", "Abuja", "Kano", "Enugu", "Ibadan",
    "Lagos", "Abuja", "Kano", "Enugu", "Lagos"
]

# Combine into a list of tuples (mini dataset)
bank_data = list(zip(names, accounts, balances, branches))

# Print dataset
print("Mini Bank Dataset (20 customers):")
for record in bank_data:
    print(record)
```

# Tuple

A tuple in Python is like a list, but it’s immutable, meaning once it’s created, you cannot change, add, or remove its elements. You use a tuple in situations where you want a collection of items that should never be modified.
