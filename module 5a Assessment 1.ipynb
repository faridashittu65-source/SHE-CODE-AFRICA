# Colour the plot

Colour the points with the variable `uses_electricity` and explain what is happening.

```python
# YOUR CODE HERE

#plt.scatter(cars_new["highway08"],cars_new["city08"], c=cars["uses_electricity"],cmap="autumn")
colors = ['red' if val == 0 else 'yellow' for val in cars["uses_electricity"]]

plt.scatter(
    cars_new["highway08"],
    cars_new["city08"],
    c=colors
)



#for the label
plt.xlabel("Highway_8")
plt.ylabel("City_8")

#Trend line
x= cars_new["highway08"]
y=cars_new["city08"]
z=np.polyfit(x,y,1)     #linear regression
p=np.poly1d(z)
plt.plot(x,p(x))
plt.show()

raise NotImplementedError()
```

From the plot, vehicles that use electricity have higher fuel efficiency values (higher city and highway MPG). This happens because electric or hybrid vehicles consume less fuel compared to traditional gasoline vehicles, so they cluster toward the upper-right part of the scatter plot, while non-electric vehicles appear lower.
