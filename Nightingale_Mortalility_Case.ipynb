```python
import pandas as pd
import numpy as np
```

```python
raw1= pd.read_excel("raw_night_1854_1855.xlsx")
```

```python
raw1.head()
```

```python
raw2 = raw1["Month"].str.split(" ", expand = True).rename(columns = {0: "Months",
                                                       1: "Year"})
raw2.head()
```

```python
raw = pd.concat([raw1, raw2], axis =1 )
raw.head()
```

```python
Raw = (raw
         .melt(["Month", "Avg Army Size","Months","Year"], 
              var_name = "Causes of Deaths", 
              value_name = " Death Count")) 
Raw.head()
```

```python
Raw["Causes of Deaths"]=Raw["Causes of Deaths"].replace("Disease Deaths","Diseases")
Raw.head()
```

```python
import janitor
from lets_plot import *
LetsPlot.setup_html()
```

```python
Raw["Sq_Death Count"] = np.sqrt(Raw[ " Death Count"])
Raw.head(5)
```

```python
Raw["Months"] = pd.Categorical(Raw["Months"],
              categories = ["Jul", "Aug", "Sep", "Oct", "Nov", "Dec", "Jan", "Feb",  "Mar", "Apr", "May", "Jun"],
              ordered = True)
```

```python
cause_order = ['Wounds', 'Other Causes', 'Diseases']
Raw["Causes of Deaths"] = pd.Categorical(Raw['Causes of Deaths'], categories=cause_order, ordered=True)
```

```python
p1 = (ggplot(Raw, aes("Months", "Sq_Death Count", fill = "Causes of Deaths"))
+ geom_bar(stat='identity', alpha = 0.6)
+ scale_fill_manual(values = ["#fb9a99", "grey30", "skyblue"])
#+ coord_equal(ratio = 1) 
+ coord_polar()
+ labs(fill = "Nature of Death",
      subtitle = "APRIL 1854 TO MARCH 1855 ") 
+ theme(axis_title_y = element_blank(),
        axis_title_x = element_blank(),
        axis_text_y = element_blank(),
        axis_ticks = element_blank(),
        plot_subtitle = element_text(hjust = 0.5),
        panel_grid = element_blank(),
        panel_border = element_blank(),
         legend_position = "bottom")
+  ggsize(900, 550))

p1
```

```python
gggrid([p2, p1], ncol = 2) + ggtitle("Diagram Of The Causes Of Mortality In The Army In The East")
```

**Interpretation of Result**:The nature of death in the army were caused by diseases rather than wounds or other causes. In the period April 1854–March 1855, disease-related deaths were extremely high. However, in April 1855–March 1856, there was a sharp reduction in deaths from diseases, indicating improved sanitary conditions. Overall, the analysis highlights how better healthcare measures greatly reduced mortality.
